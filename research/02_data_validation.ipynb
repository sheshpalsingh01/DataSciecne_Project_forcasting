```python
from dataclasses import dataclass
from pathlib import Path

@dataclass(frozen=True)
class DataValidationConfig:
    root_dir: Path
    data_file_dir: Path
    status_file: Path
    all_schema: dict
```

```python
import sys, os
from pathlib import Path
from src.forcastingproject.constants import *
from src.forcastingproject.utils.common import read_yaml_file, create_directories
from src.forcastingproject.logging.logger import logger
from src.forcastingproject.exception.exception import CustomException
```

```python
class ConfigurationManager:
    def __init__(
        self,
        config_file_path: Path = CONFIG_FILE_PATH,
        params_file_path: Path = PARAMS_FILE_PATH,
        schema_file_path: Path = SCHEMA_FILE_PATH
    ):
        self.config = read_yaml_file(config_file_path)
        self.params = read_yaml_file(params_file_path)
        self.schema = read_yaml_file(schema_file_path)

        create_directories([self.config.artifacts_root])

    def get_data_validation_config(self) -> DataValidationConfig:
        config = self.config.data_validation
        schema = self.schema.COLUMNS

        create_directories([config.root_dir])

        data_validation_config = DataValidationConfig(
            root_dir=Path(config.root_dir),
            status_file=Path(config.status_file),
            data_file_dir=Path(config.data_file_dir),
            all_schema=schema
        )
        return data_validation_config

# Backward compatibility alias
ConfigureManger = ConfigurationManager
```

```python
class DataValidation:
    def __init__(self, config: DataValidationConfig):
        self.config = config

    def validation_all_columns(self) -> bool:
        try:
            validation_status = True
            data = pd.read_csv(self.config.data_file_dir, low_memory=False)
            all_cols = list(data.columns)
            all_schema = list(self.config.all_schema.keys())

            for col in all_cols:
                if col not in all_schema:
                    validation_status = False
                    break

            with open(self.config.status_file, "w") as f:
                f.write(f"Validation status: {validation_status}\n")

            return validation_status
        except Exception as e:
            raise CustomException(e, sys)

    def get_dataframe(self) -> pd.DataFrame:
        try:
            return pd.read_csv(self.config.data_file_dir, low_memory=False)
        except Exception as e:
            raise CustomException(e, sys)

    @staticmethod
    def handle_missing_values(df: pd.DataFrame) -> pd.DataFrame:
        try:
            data = df.copy()

            # 1. CompetitionDistance -> median
            data["CompetitionDistance"] = data["CompetitionDistance"].fillna(
                data["CompetitionDistance"].median()
            )

            # 2. Competition open dates -> 0 (unknown)
            data["CompetitionOpenSinceMonth"] = (
                data["CompetitionOpenSinceMonth"].fillna(0).astype(int)
            )
            data["CompetitionOpenSinceYear"] = (
                data["CompetitionOpenSinceYear"].fillna(0).astype(int)
            )

            # 3. Promo2 features -> 0 / "None"
            data["Promo2SinceWeek"] = data["Promo2SinceWeek"].fillna(0).astype(int)
            data["Promo2SinceYear"] = data["Promo2SinceYear"].fillna(0).astype(int)
            data["PromoInterval"] = data["PromoInterval"].fillna("None")

            return data
        except Exception as e:
            raise CustomException(e, sys)
```

```python
try:
    config = ConfigurationManager()
    data_validation_config = config.get_data_validation_config()

    data_validation = DataValidation(config=data_validation_config)

    # 1) Validate schema columns
    status = data_validation.validation_all_columns()
    logger.info(f"Column validation status: {status}")

    # 2) Load dataframe and handle missing values
    df = data_validation.get_dataframe()
    df_clean = DataValidation.handle_missing_values(df)
    logger.info(f"Missing values filled. Shape: {df_clean.shape}")
    logger.info(f"Null values check after cleaning:\n{df_clean.isnull().sum()}")

except Exception as e:
    raise CustomException(e, sys)
```
